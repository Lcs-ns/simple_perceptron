```python
import numpy as np
from sklearn.datasets import load_iris
from sklearn.discriminant_analysis import StandardScaler
from sklearn.model_selection import train_test_split
```

```python
class Perceptron:
    def __init__(self, n_inputs, learning_rate=0.1, n_epochs=20):
        self.w = np.zeros(n_inputs + 1)
        self.lr = learning_rate
        self.epochs = n_epochs

    def activation(self, u):
        return 1 if u >= 0 else 0  

    def predict(self, x):
        u = np.dot(x, self.w[1:]) + self.w[0]
        return self.activation(u)

    def fit(self, X, y):
        for epoch in range(self.epochs):
            errors = 0
            for xi, target in zip(X, y):
                output = self.predict(xi)
                update = self.lr * (target - output)
                self.w[1:] += update * xi
                self.w[0] += update
                errors += int(update != 0.0)
            print(f"Época {epoch+1}/{self.epochs} - Erros: {errors}")

    def score(self, X, y):
        predictions = np.array([self.predict(xi) for xi in X])
        return np.mean(predictions == y)
```

```python
def train_binary_perceptron(target_class):
    iris = load_iris()
    X = iris.data
    y = iris.target

    # Criação do problema binário
    y_bin = np.where(y == target_class, 1, 0)

    # Padronização
    scaler = StandardScaler()
    X_scaled = scaler.fit_transform(X)

    # Split treino/teste
    X_train, X_test, y_train, y_test = train_test_split(
        X_scaled, y_bin, test_size=0.3, random_state=42
    )

    print(f"\nTreinando para: {iris.target_names[target_class]} vs demais")
    p = Perceptron(n_inputs=X.shape[1], learning_rate=0.01, n_epochs=20)
    p.fit(X_train, y_train)
    acc_train = p.score(X_train, y_train)
    acc_test = p.score(X_test, y_test)
    print(f"Acurácia treino: {acc_train:.2f}")
    print(f"Acurácia teste:  {acc_test:.2f}")
    return acc_train, acc_test

accs = []
for i in range(3):
    accs.append(train_binary_perceptron(i))

print("\nResumo final:")
for i, (train_acc, test_acc) in enumerate(accs):
    print(f"{load_iris().target_names[i]} vs demais → treino: {train_acc:.2f}, teste: {test_acc:.2f}")
```
